```python
string1 = "Hello "
string2 = input("Enter your Name: ")
string3 = string1 + string2
print(string3)
string3 = string3 + ", welcome to Python programming"
print(string3)
```

```python
print("The string is:", string3)
print("First character:", string3[0])
print("Last character:", string3[-1])
print("First 5 characters:", string3[:5])
print("Last 11 characters:", string3[-11:])
print("Reverse string:", string3[::-1])
print("Word Python:", string3[23:29])
```

```python
strM = "Python beginner tutorial"
print("Uppercase:", strM.upper())
print("Lowercase:", strM.lower())
print("Capitalize:", strM.capitalize())
print("Count of t:", strM.count('t'))
print("Replace:", strM.replace("Python", "Machine Learning"))
```

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)
t_combine = t1 + t2
print("Combined tuple:", t_combine)
print("Repeated tuple:", t_combine * 3)
print("Third element:", t_combine[2])
print("First three elements:", t_combine[:3])
print("Last three elements:", t_combine[-3:])
```
