```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)
```

A set does not allow duplicate elements. The values 'a' and 'i' appear multiple times in the given data. Python automatically removes duplicate values. Sets are unordered, so the output order may vary.

```python
my_set[4] = 's'
```

Sets are mutable, that is an object whose value or contents can be changed or modified after it is created,so we can add or remove elements. However, sets are unordered and do not support indexing. Therefore, we cannot access or modify an element using its index.To add a new element,we can use: my_set.add('s')

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
